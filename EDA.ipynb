
# Haciendo un EDA
© "El libro del DataFriendly". Jorge Moratalla


Un análisis exploratorio de datos (o EDA por sus siglas en inglés),
es un análisis general del conjunto de datos del que disponemos,
normalmente por inspección visual, para ver qué pinta tiene su distribución, obtener valores anómalos, sesgos, y otros indicadores matemáticos que nos pueden dar ciertas pistas interesantes del conjunto.
La ventaja del EDA es obtener una primera impresión, sin perder mucho tiempo, que en ocasiones es más que suficiente para tomar una decisión. No busca confirmar hipótesis específicas, sino más bien explorar y descubrir información valiosa dentro de los datos.
Para hacer un buen análisis exploratorio os recomiendo seguir una serie de pasos.


## 1. Definición del objetivo de estudio




Lo primero que tenemos que tener claro a la hora de realizar un EDA es definir bien cuál es el objetivo que se pretende obtener del dataset a analizar. Si no lo tenemos claro deberemos tener reuniones con la Alta Dirección o aquellos interlocutores que nos permitan decidir el tipo de estudio a realizar y qué es lo que se pretende del mismo.

Es posible que, como consecuencia de este estudio, se obtengan algunos hallazgos que inicialmente no estaban contemplados, pero que nos pueden ayudar a identificar posibles puntos de mejora que siempre son bien acogidos.

Tomando el dataset del ejemplo que vamos a estudiar, podremos definir el objetivo inicial como la identificación de las posibles causas que está provocando la pérdida de talento, la falta de retención de personal en algunas categorías, y el elevado índice de rotación que presenta la compañía. Además se deberá investigar dentro del objeto de estudio cual es la mejor forma de gestionar las contrataciones a realizar o las acciones a llevar a cabo para mejorar la relación con la empresa y evitar las posibles pérdidas económicas que está causando esta crisis.

##2. Primera aproximación a las medidas estadísticas del dataset

Lo primero que deberemos hacer para tener una aproximación de los datos a los cuales nos enfrentamos es conocer las principales medidas estadísticas, que nos van a determinar de partida la tendencia, la dispersión y los valores extremos. Vamos a repasar algunas nociones de estadística para comprender qué nos puede aportar cada una de ellas y cómo podemos aprovecharlas en nuestro beneficio.

En estadística llamamos población al conjunto de elementos que forman parte de un estudio. Como probablemente no es posible analizar todos y cada uno de los elementos, trabajaremos con una muestra, es decir, un conjunto representativo de la población. A cada miembro de la muestra lo denominamos individuo, cada uno de ellos con un conjunto de variables a estudiar

Para facilitar el estudio de la muestra, vamos a intentar resumir el conjunto de observaciones en unos pocos números, es lo que se denomina tendencia central. Para ello utilizaremos las siguientes medidas:

* Tamaño de la muestra: determina el número de observaciones y características

* Valores mínimo y máximo de la muestra

* Media aritmética: permite resumir un conjunto de datos en un único valor representativo

* Mediana: proporciona un punto medio de los valores de nuestra variable

* Moda: identifica el valor más frecuente

* Rango: es la diferencia entre el valor máximo y mínimo

* Desviación típica: indica cuánto se alejan, en promedio, los valores individuales de la variable de su valor medio

* Varianza: a partir de la desviación típica podemos obtener la varianza, que indica cómo de dispersos están los valores de la variable. Una varianza baja sugiere que los datos están agrupados cerca de la media, mientras que una varianza alta indica que los datos están más dispersos.

* Cuartiles y rango intercuartil (RIC): los cuartiles dividen un conjunto de datos en cuatro partes iguales, mientras que el RIC, calculado como la diferencia entre el tercer y primer cuartil, representa la dispersión del 50% central de los datos, siendo útil para identificar valores atípicos y comprender la variabilidad de la muestra. Vamos a ver un ejemplo de esto. Imaginemos que tenemos una muestra de valores de una variable calificaciones. Calculamos la mediana que corresponde con el cuartil Q2 (esto quiere decir que el 50% de las calificaciones están por debajo de él). A partir de ahí obtenemos Q1 y Q3 obteniendo la mediana con las franjas que nos quedan.

##4. Determinar la distribución de las variables

La forma más visual de determinar el aspecto de las distintas variables que conforman nuestro conjunto de datos es pintando la distribución de cada una de ellas, siempre que sea posible. Una de las distribuciones de datos clásicas es la distribución normal, que se caracteriza por tener una forma de campana de Gauss, concentrando la mayoría de las frecuencias en la zona central de la tabla. Es la distribución más frecuente en el mundo de los datos, como por ejemplo la altura de las personas, la renta per cápita, las calificaciones de exámenes, etc.

Una distribución normal nos sirve para predecir de una forma muy simple, mediante cálculo de probabilidades a través de unas tablas estándar. Se forma colocando en el eje de abscisas las observaciones y el eje de ordenadas la frecuencia, jugando con dos parámetros principales: la media (μ) y la desviación estándar (σ). La media indica el centro de la distribución, mientras que la desviación estándar determina su dispersión o anchura.



image.png

Para que una distribución sea normal se deben dar estas condiciones:
* 50% de las observaciones están por encima de la media
* 50% de las observaciones están por debajo de la media
* 68% de las observaciones están comprendidas en el intervalo que va desde la media menos la desviación estándar hasta la media más la desviación estándar
* 95% de las observaciones están comprendidas en el intervalo que va desde la media menos dos veces la desviación estándar hasta la media más 2 veces la desviación estándar
* 99,7% de las observaciones están comprendidas en el intervalo que va desde la media menos tres veces la desviación estándar hasta la media más 3 veces la desviación estándar

Una vez tengamos esta distribución es muy fácil hacer una predicción en base a las probabilidades, en base a la tabla de la distribución normal estándar o tabla Z. Veámoslo con un ejemplo. Imagina que la variable “calificación de alumno” sigue una distribución normal de media 5 y desviación típica 0,5. Si queremos calcular la probabilidad de que un alumno obtenga una calificación superior a 6,56 se calcula de la siguiente forma:

Otra de las distribuciones más frecuentes que nos podemos encontrar es la denominada long-tail, que se caracteriza por distribuir de forma ordenada las frecuencias de un gran conjunto de valores que puede tomar las variables categóricas. A la izquierda encontraremos un conjunto de valores reducidos que toman frecuencias elevadas, y a medida que nos desplazamos a la derecha van disminuyendo de forma exponencial. Sin embargo este tipo de distribución no permite una inferencia probabilística como ofrece la normal.

El modelo no está mal vamos a ver si poemos afinarlo más, para ello vamos a quedarnos sólo con las variables que más correlan

````python
#Selección de variables con eliminación regresiva
import statsmodels.api as sm
def backward_elimination(X, y, significance_level=0.05):
  cols = list(X.columns)
  while(True):
    p_values = sm.Logit(y, X[cols]).fit().pvalues
    #Identificamos la variable con el p-valor más alto o la que menos contribuye al modelo.
    max_p_value = p_values.max()
    #Si ese valor excede un umbral (por ejemplo, p > 0.05), la eliminas
    if max_p_value > significance_level:
      excluded_col = p_values.idxmax()
      cols.remove(excluded_col)
    else:
      #Repites hasta que todas las variables restantes sean estadísticamente significativas.
      break
  return X[cols]
````

````python
#divididimos conjunto en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, shuffle=False)

X_selected = backward_elimination(X_train, y_train)
X_test_selected = X_test[X_selected.columns]
print(X_selected)

# Crear y entrenar el modelo
model = LogisticRegression()
model.fit(X_selected, y_train)

# Evaluar el modelo
model.score(X_test_selected, y_test)
````

````python
import numpy as np
def forward_selection_aic(data, target_col):
    remaining_vars = data.columns.tolist()
    remaining_vars.remove(target_col)
    selected_vars = []
    current_aic = np.inf
    while remaining_vars:
        aic_with_var_added = []
        for var in remaining_vars:
            vars_to_test = selected_vars + [var]
            X_train = sm.add_constant(data[vars_to_test])
            y_train = data[target_col]
            model = sm.Logit(y_train, X_train).fit(disp=0)
            aic_with_var_added.append((var, model.aic))
        # Buscar la variable que más baje el AIC
        best_var, best_aic = min(aic_with_var_added, key=lambda x: x[1])

        if best_aic < current_aic:
            selected_vars.append(best_var)
            remaining_vars.remove(best_var)
            current_aic = best_aic
        else:
            break
    return selected_vars
#Ejecutar forward selection
vars_finales = forward_selection_aic(df, "LeaveOrNot")
print("Variables seleccionadas por AIC (hacia adelante):", vars_finales)
````

````python
#divididimos conjunto en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X[vars_finales],y,test_size=0.2, shuffle=False)


# Crear y entrenar el modelo
model = LogisticRegression()
model.fit(X_train, y_train)

# Evaluar el modelo
model.score(X_test, y_test)
````

````python
#Escalamos las variables temp_min, temp_max, velocidad_viento, indice_ultravioleta
mms = MinMaxScaler()
df[['temp_min','temp_max','velocidad_viento','Indice_ultravioleta']] = mms.fit_transform(df[['temp_min','temp_max','velocidad_viento','Indice_ultravioleta']])
display(df)
````

````python
#Transformamos el campo Lluvia a 1 si mayor que 75% 0 en otro caso y calculamos dumies
df['Lluvia'] = df['Lluvia'].apply(lambda x: 0 if x < 75 else 1)

direccion_viento = pd.get_dummies(df['dir_viento'], prefix = 'dir_viento').astype(int)
df = pd.concat([df['temp_min'],df['temp_max'],df['velocidad_viento'],df['Indice_ultravioleta'],direccion_viento,df['Lluvia']],axis = 1)
df
````

````python
#Obtenemos las variables que correlan con Lluvia en más de un 0.6
variables = []
c = df.corr()
variables = c[(c['Lluvia'] > 0.6)].drop('Lluvia', axis=0).index
print(variables)

#formamos un diccionario
lista_dic = []
for elem in variables:
  dic = {'variable':elem,'correlacion':float(c[elem]['Lluvia'])}
  lista_dic.append(dic)

print(lista_dic)
````

````python
#entrenamos un modelo
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

X = df[variables]
y = df["Lluvia"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=False)

modelo = LogisticRegression()
modelo.fit(X_train,y_train)
modelo.score(X_test,y_test)
````
